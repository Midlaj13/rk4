```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
a=0
b=2
n=400000
h=(b-a)/n
domain=np.linspace(a,b,n)
```

```python
def fn(x,y):
    return 3*math.log(1+math.sin(x))/(math.sin(x)+3)
```

```python
def k1(x,y):
    return h*fn(x,y)
```

```python
def k2(x,y):
    return h*fn((x+(h/2)),(y+(k1(x,y)/2)))
```

```python
def k3(x,y):
    return h*fn((x+(h/2)),(y+(k2(x,y)/2)))
```

```python
def k4(x,y):
    return h*fn((x+(h/2)),(y+(k3(x,y))))
```

```python
result=np.zeros(n)
result[0]=5
```

```python
for i in range(1,n):
    result[i]=result[i-1]+(k1(domain[i-1],result[i-1])/6)+(k2(domain[i-1],result[i-1])/3)+(k3(domain[i-1],result[i-1])/3)+(k4(domain[i-1],result[i-1])/6)
```

```python
plt.plot(domain,result)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
